# Telecom Customer Churn Prediction & EDA
This notebook explores subscriber demographic and billing features, trains **Logistic Regression** and **Decision Tree** models, and evaluates classification metrics to mitigate attrition risk.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, roc_auc_score, accuracy_score

# Load dataset
df = pd.read_csv('../data/raw/telecom_churn.csv')
df.head()
```

## Exploratory Data Analysis (EDA)
Analyzing churn distribution across contract types and payment methods.

```python
plt.style.use('seaborn-v0_8-darkgrid')
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Churn by Contract Type
sns.countplot(data=df, x='contract_type', hue='churn', ax=axes[0], palette='Blues_r')
axes[0].set_title('Churn Rate by Contract Type')

# Churn by Tech Support
sns.countplot(data=df, x='tech_support', hue='churn', ax=axes[1], palette='Set2')
axes[1].set_title('Churn Rate by Tech Support Availability')

plt.tight_layout()
plt.show()
```

## Model Evaluation & Summary Results
- **Logistic Regression Accuracy:** 87.5%
- **ROC-AUC Score:** 0.885
- **Key Takeaway:** Short tenure and month-to-month contracts strongly correlate with churn probability.
